# Monte Carlo CBA post-processing
Projects, hierarchy, costs and cost shares are declared in `config/projects.yaml`. SEW = reference - counterfactual, paired per sample.

```python
import pandas as pd
from montecarlo import cba

CONFIG = "config/projects.yaml"
OUTDIR = None  # e.g. "results/example-50-discrete/cba" to also write CSVs
pd.options.display.float_format = "{:,.2f}".format
```

```python
tables = cba.run(CONFIG, outdir=OUTDIR)
sew = tables["cba_sew_samples"]
print(sew.project.nunique(), "projects,", sew["sample"].nunique(), "samples,", sew.country.nunique(), "countries")
sew.head()
```

## SEW summary and per-country ranking
`rank_mean`: 1 = highest expected SEW. `rank_p5`: 1 = best downside (P5). `rank_std`: 1 = lowest spread.

```python
summary = tables["cba_summary_sew"]
summary
```

```python
# Expected SEW matrix: projects x countries
summary.pivot(index="project", columns="country", values="mean")
```

```python
# Rank matrix per country (by expected SEW)
summary.pivot(index="project", columns="country", values="rank_mean")
```

## Net benefit (SEW minus the cost share borne by each country)
Only for projects with `cost` and `cost_share` in the YAML.

```python
tables.get("cba_summary_net_benefit", "No project has cost information")
```

## Synergy of bundles
SEW(bundle) - sum of SEW(children), per sample and country.
